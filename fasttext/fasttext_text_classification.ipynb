```python
import pandas as pd
import numpy as np

df = pd.read_csv('ecommerce_dataset.csv', names = ['category','discription'], header = None)
print(df.shape)
df.head()
```

```python
df.dropna(inplace=True)
df.shape
```

```python
df.category.value_counts()
```

```python
df.category.replace("Clothing & Accessories", "Clothing_Accessories", inplace= True)
```

```python
df.category.unique()
```

```python
df.category = '__label__' + df.category
df.head()
```

```python
df['category_description'] = df.category + " " + df.discription
df.head()
```

```python
import re
text = "  VIKI's | Bookcase/Bookshelf (3-Shelf/Shelve, White) | ? . hi"
text = re.sub(r'[^\w\s\']', ' ', text)
text = re.sub(' +', ' ',text)
text.strip().lower()
```

```python
def preprocess(text):
    text = re.sub(r'[^\w\s\']', ' ', text)
    text = re.sub(' +', ' ', text)
    return text.strip().lower()
```

```python
df['category_description'] = df.category_description.map(preprocess)
df.head()
```

```python
df.category_description[0]
```

```python
from sklearn.model_selection import train_test_split
train , test = train_test_split(df,test_size=0.2)
```

```python
train.shape , test.shape
```

```python
train.to_csv('ecommerce.train', columns=['category_description'], index= False, header=False)
test.to_csv('ecommerce.test', columns=['category_description'], index= False, header=False)
```

```python
import fasttext
model = fasttext.train_supervised('ecommerce.train')
model.test('ecommerce.test')
```
